## 1. `for`: recorrer elemento a elemento

En C o Java, el `for` clásico es un **contador**: `for (int i = 0; i < n; i++)`.

En Python, `for` es un **"para cada"** (*for each*): recorre los elementos de cualquier objeto **iterable** (cadenas, rangos, listas, ficheros…) y en cada vuelta la variable toma el valor del siguiente elemento.

```python
for elemento in iterable:
    bloque
```

````python
for letra in "Python":
    print(letra)
````

````python
palabra = "murciélago"
vocales = 0

for letra in palabra:
    if letra in "aeiouáéíóú":
        vocales += 1

print(f"'{palabra}' tiene {vocales} vocales")
````

## 2. `range()`: secuencias de números

`range` genera una secuencia de enteros. Es la forma de hacer en Python el bucle "contador" de otros lenguajes.

| Llamada | Genera |
| --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 (el final **no se incluye**) |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 3)` | 0, 3, 6, 9 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(10, 0, -2):
    print(i, end=" ")
print()

print(sum(range(1, 101)))      # suma de 1 a 100
````

## 3. `enumerate()` y `zip()`: adiós a los índices

Quien viene de C tiende a escribir `for i in range(len(texto)):` para tener el índice. En Python hay herramientas mejores:

- **`enumerate(iterable, start=0)`**: da en cada vuelta la **posición y el elemento**.
- **`zip(a, b)`**: recorre **varios iterables a la vez**, emparejando sus elementos.

````python
for posicion, letra in enumerate("hola", start=1):
    print(f"Letra {posicion}: {letra}")
````

````python
for letra, numero in zip("abc", "123"):
    print(letra + numero, end=" ")
````

### 🧩 Reto 3.1 · El estilo C en Python  ⭐

Las dos celdas hacen lo mismo. ¿Por qué la segunda se considera mejor? Piensa en legibilidad y en qué pasaría si el objeto recorrido no tuviera índices.

````python
texto = "abc"
for i in range(len(texto)):
    print(i, texto[i])
````

````python
texto = "abc"
for i, letra in enumerate(texto):
    print(i, letra)
````

## 4. `while`: repetir mientras se cumpla una condición

Se usa cuando **no sabemos de antemano cuántas vueltas** habrá que dar. La condición se comprueba **antes** de cada vuelta.

```python
while condición:
    bloque
```

⚠️ Algo dentro del bloque tiene que hacer que la condición acabe siendo falsa. Si no, el bucle es **infinito**.

````python
capital = 1000
anios = 0

while capital < 2000:
    capital *= 1.05          # 5 % de interés anual
    anios += 1

print(f"El capital se duplica en {anios} años ({capital:.2f} €)")
````

Patrón muy habitual: **`while True` + `break`**, cuando la condición de salida se conoce dentro del bucle. Por ejemplo, pedir datos hasta que sean válidos (no lo ejecutamos porque `input()` espera al teclado):

```python
while True:
    texto = input("Edad: ")
    if texto.isdigit():
        edad = int(texto)
        break
    print("Introduce un número entero")
```

Python **no tiene `do ... while`**; este patrón es su sustituto.

## 5. `break`, `continue` y `else`

| Instrucción | Efecto |
| --- | --- |
| `break` | Sale del bucle inmediatamente |
| `continue` | Salta al principio de la siguiente vuelta |
| `else` (del bucle) | Se ejecuta si el bucle termina **sin** haber hecho `break` |

````python
for letra in "Python es genial":
    if letra == " ":
        continue          # nos saltamos los espacios
    if letra == "g":
        break             # paramos al llegar a la g
    print(letra, end="")
````

### 🧩 Reto 5.1 · El `else` de los bucles  ⭐⭐

Casi nadie conoce el `else` de los bucles. Ejecuta con `numero = 91` y con `numero = 97`. ¿Cuándo se ejecuta el `else`? ¿Qué ventaja tiene frente a usar una variable `encontrado = False`?

````python
numero = 91

for divisor in range(2, numero):
    if numero % divisor == 0:
        print(f"{numero} no es primo: es divisible entre {divisor}")
        break
else:
    print(f"{numero} es primo")
````

## 6. Bucles anidados

Un bucle dentro de otro: por cada vuelta del exterior, el interior se ejecuta completo.

````python
for fila in range(1, 6):
    for columna in range(1, 6):
        print(f"{fila * columna:4}", end="")
    print()
````

### 🧩 Reto 6.1 · ¿Cuántas vueltas?  ⭐⭐

````python
n = 100
contador = 0

for i in range(n):
    for j in range(n):
        contador += 1

print(contador)
````

## ✅ Resumen

- El `for` de Python es un **"para cada"**: recorre los elementos de un iterable. Modificar la variable no altera la iteración.
- La variable del bucle **sigue existiendo después**… salvo que el iterable estuviera vacío.
- `range(inicio, fin, paso)`: el fin **no se incluye**, solo admite enteros y es **perezoso** (no ocupa memoria).
- Evita `for i in range(len(x))`: usa **`enumerate`**. Para recorrer varias cosas a la vez, **`zip`** (con `strict=True` si deben tener la misma longitud).
- `while` para cuando no sabes cuántas vueltas habrá. **Nunca** uses `!=` con `float` como condición de parada.
- `while True` + `break` es el `do ... while` de Python.
- `break` sale solo del bucle **más interno**. El `else` de un bucle se ejecuta si **no** hubo `break`.
- Bucles anidados = coste **cuadrático**: cuidado con muchos datos.
