### Métodos principales

| Método | Qué hace | Devuelve |
| --- | --- | --- |
| `append(x)` | Añade `x` al final | `None` |
| `extend(iterable)` | Añade todos los elementos del iterable | `None` |
| `insert(i, x)` | Inserta `x` en la posición `i` | `None` |
| `pop(i=-1)` | Quita y **devuelve** el elemento de la posición `i` | el elemento |
| `remove(x)` | Quita la **primera** aparición de `x` | `None` |
| `index(x)` | Posición de la primera aparición de `x` | `int` |
| `count(x)` | Cuántas veces aparece `x` | `int` |
| `sort()` | Ordena **la propia lista** | `None` |
| `reverse()` | Invierte **la propia lista** | `None` |

````python
frutas = ["manzana", "pera"]
frutas.append("uva")
frutas.insert(0, "kiwi")
frutas.extend(["melón", "fresa"])
print(frutas)

ultima = frutas.pop()
print(ultima, frutas)

frutas.remove("pera")
frutas[0] = "KIWI"               # las listas son mutables
print(frutas)

print(frutas.index("uva"), frutas.count("uva"))
````

## 3. Comprensiones

Una **comprensión** crea una colección nueva a partir de un iterable, en una sola línea. Es una de las características más idiomáticas de Python.

```python
nueva = [expresión for elemento in iterable if condición]
```

````python
cuadrados = []
for x in range(10):
    cuadrados.append(x ** 2)
print(cuadrados)

cuadrados = [x ** 2 for x in range(10)]      # lo mismo en una línea
print(cuadrados)

pares = [x for x in range(20) if x % 2 == 0]
print(pares)

palabras = ["Hola", "MUNDO", "pYtHoN"]
print([p.lower() for p in palabras])
````

Si se escribe entre **paréntesis** en lugar de corchetes, es una **expresión generadora**: no crea la lista, sino que genera los valores uno a uno bajo demanda (como `range`, notebook 05). Ideal para pasarla directamente a `sum()`, `max()`, `any()`, `all()`…

````python
print(sum(x ** 2 for x in range(1_000_000)))
print(any(x > 100 for x in [5, 50, 500]))     # ¿alguno cumple?
print(all(x > 1 for x in [5, 50, 500]))       # ¿todos cumplen?
````

## 5. Diccionarios

Un diccionario guarda pares **clave → valor**. Se accede por clave, no por posición. Es la estructura más importante de Python: la usa internamente el propio lenguaje, y un JSON es, en la práctica, diccionarios y listas anidados.

````python
persona = {
    "nombre": "Ana",
    "edad": 34,
    "idiomas": ["español", "inglés"],
}

print(persona["nombre"])
print(persona["idiomas"][1])

persona["edad"] = 35              # modificar
persona["ciudad"] = "Vigo"        # añadir
del persona["idiomas"]            # borrar
print(persona)

print("edad" in persona)          # in busca en las CLAVES
print(len(persona))
````

### 🧩 Reto 6.2 · ¿Dónde buscar es más rápido?  ⭐⭐

Buscamos el mismo número en una lista y en un conjunto de un millón de elementos. ¿Por qué hay tanta diferencia de tiempo?

````python
import time

numeros_lista = list(range(1_000_000))
numeros_set = set(numeros_lista)

inicio = time.perf_counter()
for _ in range(100):
    999_999 in numeros_lista
t_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
for _ in range(100):
    999_999 in numeros_set
t_set = time.perf_counter() - inicio

print(f"lista: {t_lista:.4f} s")
print(f"set:   {t_set:.6f} s")
print(f"El set es unas {t_lista / t_set:,.0f} veces más rápido")
````

## 8. `match` con estructuras

En el notebook 04 dijimos que `match` brilla con estructuras. Puede comprobar la **forma** de una lista o un diccionario y **extraer** sus partes a la vez:

````python
mensajes = [
    {"tipo": "error", "codigo": 404},
    {"tipo": "aviso", "texto": "Disco al 90 %"},
    ["mover", 10, 20],
    ["salir"],
    {"tipo": "desconocido"},
]

for mensaje in mensajes:
    match mensaje:
        case {"tipo": "error", "codigo": codigo}:
            print(f"Error con código {codigo}")
        case {"tipo": "aviso", "texto": texto}:
            print(f"Aviso: {texto}")
        case ["mover", x, y]:
            print(f"Mover a ({x}, {y})")
        case ["salir"]:
            print("Salir")
        case _:
            print(f"No reconocido: {mensaje}")
````

- Un patrón de **lista** (`["mover", x, y]`) encaja con secuencias de **exactamente** esa longitud, y captura las posiciones en variables.
- Un patrón de **diccionario** (`{"tipo": "error", "codigo": codigo}`) encaja si **tiene esas claves** (puede tener más) y captura sus valores.

Es muy útil para procesar mensajes de una API, comandos o eventos.

## ✅ Resumen

| Necesito… | Uso |
| --- | --- |
| Una secuencia ordenada que voy a modificar | `list` |
| Un registro fijo, o una clave compuesta | `tuple` |
| Buscar valores por un nombre o identificador | `dict` |
| Eliminar duplicados o comprobar pertenencia rápido | `set` |

- Índices y rebanadas igual que en las cadenas. Una rebanada fuera de rango no da error.
- `append` añade un elemento; `extend`, varios. `sort()` devuelve `None`; `sorted()`, una lista nueva.
- **Nunca** cambies el tamaño de una lista o un diccionario mientras lo recorres.
- La asignación **no copia**. `.copy()` hace una copia **superficial**; para estructuras anidadas, `copy.deepcopy()`.
- `[[0] * 3] * 3` repite **la misma** fila: usa una comprensión.
- Comprensiones para construir colecciones de forma sencilla; si se complican, bucle normal.
- Diccionarios: `get()` para claves que pueden faltar. Las claves deben ser **hashables** (no listas).
- `{}` es un diccionario vacío; el conjunto vacío es `set()`.
- `in` sobre `set`/`dict` es O(1); sobre `list`, O(n).
