## 1. Definir y llamar funciones

```python
def nombre_funcion(parametro1, parametro2):
    """Docstring: qué hace la función."""
    ...
    return resultado
```

- `def` **crea un objeto función** y lo asigna al nombre indicado.
- `return` termina la función y **devuelve** un valor a quien la llamó.
- Si no hay `return` (o es un `return` sin valor), la función devuelve **`None`**.

````python
def area_triangulo(base, altura):
    """Devuelve el área de un triángulo."""
    return base * altura / 2


resultado = area_triangulo(10, 4)
print(resultado)
print(area_triangulo(3, 3) + area_triangulo(2, 2))
````

Una función puede devolver **varios valores**: en realidad devuelve una **tupla**, que se puede desempaquetar (notebook 06).

````python
def estadisticas(numeros):
    """Devuelve el mínimo, el máximo y la media de una lista de números."""
    return min(numeros), max(numeros), sum(numeros) / len(numeros)


minimo, maximo, media = estadisticas([4, 8, 15, 16, 23, 42])
print(minimo, maximo, media)
print(type(estadisticas([1, 2])))
````

## 2. Parámetros y argumentos

- **Parámetro**: el nombre en la definición (`def f(x)`).
- **Argumento**: el valor en la llamada (`f(5)`).

| Tipo | Definición | Llamada |
| --- | --- | --- |
| Posicional | `def f(a, b)` | `f(1, 2)` |
| Por nombre (*keyword*) | `def f(a, b)` | `f(b=2, a=1)` |
| Valor por defecto | `def f(a, b=10)` | `f(1)` |
| Número variable de posicionales | `def f(*args)` | `f(1, 2, 3)` → `args = (1, 2, 3)` |
| Número variable por nombre | `def f(**kwargs)` | `f(x=1, y=2)` → `kwargs = {"x": 1, "y": 2}` |
| Solo por nombre | `def f(a, *, b)` | `f(1, b=2)` (obligatorio nombrar `b`) |
| Solo posicional | `def f(a, /, b)` | `f(1, 2)` o `f(1, b=2)` (prohibido nombrar `a`) |

````python
def saludar(nombre, saludo="Hola", signo="!"):
    return f"{saludo}, {nombre}{signo}"


print(saludar("Ana"))
print(saludar("Luis", "Buenos días"))
print(saludar("Eva", signo="."))
print(saludar(signo="?", nombre="Marta"))
````

````python
def sumar_todo(*numeros):
    print("numeros =", numeros)
    return sum(numeros)


def mostrar_ficha(**datos):
    for clave, valor in datos.items():
        print(f"{clave}: {valor}")


print(sumar_todo(1, 2, 3, 4))
mostrar_ficha(nombre="Ana", edad=34, ciudad="Vigo")

valores = [10, 20, 30]
print(sumar_todo(*valores))           # * desempaqueta una lista en argumentos
````

## 3. Ámbito de las variables (*scope*)

Cada llamada a una función tiene su propio **espacio de nombres local**. Al buscar un nombre, Python sigue la regla **LEGB**:

1. **L**ocal: dentro de la función actual.
2. **E**nclosing: en las funciones que la contienen (si está anidada).
3. **G**lobal: en el módulo (en un notebook, las variables de las celdas).
4. **B**uilt-in: los nombres integrados (`print`, `len`…).

````python
mensaje = "global"


def mostrar():
    mensaje = "local"
    print("Dentro:", mensaje)


mostrar()
print("Fuera: ", mensaje)
````

## 4. Las funciones son objetos

Una función es un objeto como cualquier otro: se puede asignar a una variable, guardar en una lista o en un diccionario, **pasar como argumento** y **devolver** desde otra función.

````python
def gritar(texto):
    return texto.upper() + "!"


def susurrar(texto):
    return texto.lower() + "..."


def aplicar(funcion, valor):
    return funcion(valor)


print(aplicar(gritar, "Hola"))
print(aplicar(susurrar, "Hola"))

estilos = {"alto": gritar, "bajo": susurrar}
print(estilos["bajo"]("ADIÓS"))
````

### Funciones como argumento: `key=`

Ya lo usamos en el notebook 06 con `sorted(..., key=len)` y `max(..., key=dict.get)`. La función `key` se aplica a cada elemento para decidir cómo ordenarlo.

Para funciones cortas de un solo uso existe **`lambda`**: una función anónima de **una sola expresión**.

````python
personas = [("Ana", 34), ("Luis", 28), ("Eva", 41)]

print(sorted(personas, key=lambda persona: persona[1]))
print(sorted(personas, key=lambda persona: persona[0], reverse=True))
print(max(personas, key=lambda persona: persona[1]))

doble = lambda x: x * 2         # funciona, pero PEP 8 lo desaconseja: usa def
print(doble(21))
````

### *Closures*: funciones que recuerdan

Una función definida **dentro de otra** puede usar las variables de la función exterior, **incluso después de que esta haya terminado**. Eso es un *closure* (clausura). Para **modificar** esas variables se usa `nonlocal`.

````python
def crear_multiplicador(factor):
    def multiplicar(x):
        return x * factor          # factor viene del ámbito exterior (E de LEGB)
    return multiplicar


doble = crear_multiplicador(2)
triple = crear_multiplicador(3)
print(doble(10), triple(10))
````

### Decoradores (introducción)

Un **decorador** es una función que recibe una función y devuelve otra que la "envuelve" para añadirle comportamiento. La sintaxis `@decorador` encima de un `def` es solo un atajo para `funcion = decorador(funcion)`.

Los verás constantemente en librerías: `@app.get("/")` en FastAPI, `@pytest.fixture`, `@property`, `@functools.cache`…

````python
import time
import functools


def cronometrar(funcion):
    @functools.wraps(funcion)              # conserva el nombre y el docstring originales
    def envoltura(*args, **kwargs):
        inicio = time.perf_counter()
        resultado = funcion(*args, **kwargs)
        print(f"{funcion.__name__} tardó {time.perf_counter() - inicio:.4f} s")
        return resultado
    return envoltura


@cronometrar
def suma_cuadrados(n):
    return sum(x * x for x in range(n))


print(suma_cuadrados(1_000_000))
````

## 5. Recursividad

Una función **recursiva** se llama a sí misma. Necesita siempre un **caso base** que detenga la recursión.

````python
def factorial(n):
    """n! calculado de forma recursiva."""
    if n <= 1:          # caso base
        return 1
    return n * factorial(n - 1)


print(factorial(5))
print(factorial(30))
````

````python
import sys

print("Límite de recursión:", sys.getrecursionlimit())

try:
    factorial(5000)
except RecursionError as e:
    print(f"RecursionError: {e}")
````

### 🧩 Reto 5.2 · Fibonacci lento  ⭐⭐⭐

Para calcular `fibonacci(30)` la función se llama casi **3 millones de veces**, y `fibonacci(40)` necesitaría más de **300 millones** (decenas de segundos). ¿Por qué tantas? Fíjate en el contador de llamadas. Después ejecuta la versión con `@functools.cache`.

````python
import time

llamadas = 0


def fibonacci(n):
    global llamadas
    llamadas += 1
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)


inicio = time.perf_counter()
print(fibonacci(30))
print(f"{llamadas:,} llamadas en {time.perf_counter() - inicio:.2f} s")
````

````python
import functools


@functools.cache
def fibonacci_rapido(n):
    if n < 2:
        return n
    return fibonacci_rapido(n - 1) + fibonacci_rapido(n - 2)


inicio = time.perf_counter()
print(fibonacci_rapido(300))
print(f"{time.perf_counter() - inicio:.5f} s")
print(fibonacci_rapido.cache_info())
````

## ✅ Resumen

- `def` es una instrucción que **crea un objeto función** al ejecutarse. Sin `return`, la función devuelve `None`.
- **Devuelve** los resultados; no los imprimas dentro de la función.
- Argumentos: posicionales, por nombre, por defecto, `*args`, `**kwargs`, solo-nombre (`*`) y solo-posición (`/`).
- Los valores por defecto se evalúan **una vez**: **nunca uses un mutable** como valor por defecto (usa `None`).
- Paso **por asignación**: la función puede modificar los mutables que recibe, pero no reasignar las variables de fuera.
- Regla **LEGB**. Si asignas un nombre en una función, es **local en toda la función** (`UnboundLocalError`). Evita `global`.
- Las funciones son objetos: se pasan como argumento (`key=`), `lambda` para funciones de una expresión.
- Las lambdas y *closures* buscan las variables **al ejecutarse** (*late binding*).
- Decoradores: `@decorador` equivale a `f = decorador(f)`. Ejemplo útil: `@functools.cache`.
- Recursividad: límite de 1000 niveles y sin optimización de cola. Muchas veces, mejor un bucle.
