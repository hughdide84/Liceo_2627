## 1. `if`: ejecutar un bloque solo si se cumple una condición

```python
if condición:
    bloque           # se ejecuta solo si la condición es verdadera
siguiente_instrucción
```

- La línea del `if` termina en **`:`**.
- El bloque es todo lo que tiene **más sangría** (notebook 01).
- La condición puede ser **cualquier expresión**: Python aplica la *truthiness* (notebook 02).

````python
temperatura = 31

if temperatura > 30:
    print("Hace mucho calor")
    print("Recuerda beber agua")

print("Fin del programa")
````

Gracias a la *truthiness*, no hace falta comparar con `""`, `0` o `None` de forma explícita:

````python
nombre = ""

if not nombre:
    print("No has indicado tu nombre")

if nombre == "":          # equivalente, pero menos idiomático
    print("No has indicado tu nombre")
````

## 2. `else` y `elif`

- `else`: lo que se ejecuta si **ninguna** condición anterior se cumple.
- `elif` (*else if*): una condición alternativa que **solo se comprueba si las anteriores han fallado**.
- Python ejecuta **como máximo un bloque** de toda la cadena: el de la primera condición verdadera.

````python
nota = 7.5

if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(f"{nota} -> {calificacion}")
````

💡 Fíjate en que en el segundo `elif` no hace falta escribir `nota >= 7 and nota < 9`: si hemos llegado ahí, ya sabemos que la nota es menor que 9.

## 3. La expresión condicional (operador ternario)

Cuando solo queremos **elegir entre dos valores**, existe una forma compacta:

```python
valor = valor_si_verdadero if condición else valor_si_falso
```

Es una **expresión** (produce un valor), así que se puede usar en una asignación, dentro de un `print` o de una f-string. El orden es distinto al `cond ? a : b` de C, Java o JavaScript: en Python se lee casi como una frase.

````python
numero = 7
tipo = "par" if numero % 2 == 0 else "impar"
print(f"{numero} es {tipo}")

stock = 0
print(f"Estado: {'disponible' if stock > 0 else 'agotado'}")
````

## 4. `match` / `case` (Python 3.10+)

`match` compara un valor con una serie de **patrones** y ejecuta el primero que encaja. Parece un `switch`, pero es mucho más potente (*structural pattern matching*). De momento veremos lo básico; cuando conozcamos las colecciones y las clases veremos patrones más avanzados.

````python
comando = "salir"

match comando:
    case "ayuda":
        print("Mostrando ayuda...")
    case "salir" | "exit" | "q":          # varias opciones con |
        print("Cerrando el programa")
    case _:                               # comodín: cualquier otro valor
        print(f"Comando desconocido: {comando}")
````

Se puede añadir una **guarda** (`if`) a un patrón, y **capturar** el valor en una variable:

````python
temperatura = -3

match temperatura:
    case 0:
        print("Justo en el punto de congelación")
    case t if t < 0:
        print(f"{t} °C: bajo cero")
    case t if t > 35:
        print(f"{t} °C: calor extremo")
    case _:
        print("Temperatura normal")
````

## 5. El operador morsa `:=`

`:=` asigna **y** devuelve el valor, así que se puede usar dentro de una condición. Evita calcular dos veces lo mismo:

````python
texto = "Python es un lenguaje muy versátil"

if (longitud := len(texto)) > 20:
    print(f"Texto demasiado largo: {longitud} caracteres (máximo 20)")
````

## 6. Simplificar condiciones

Algunas reglas útiles:

| En lugar de… | Mejor… |
| --- | --- |
| `if x == True:` | `if x:` |
| `if len(texto) == 0:` | `if not texto:` |
| `if x == None:` | `if x is None:` |
| `if edad >= 18 and edad < 65:` | `if 18 <= edad < 65:` |
| `if not (a and b):` | `if not a or not b:` (leyes de De Morgan) |
| `if cond: x = True` / `else: x = False` | `x = cond` |

## ✅ Resumen

- `if` / `elif` / `else`: se ejecuta **como máximo un bloque**, el de la **primera** condición verdadera. **El orden importa.**
- Varios `if` seguidos son decisiones **independientes**; `elif` las hace **excluyentes**.
- Aprovecha la *truthiness*: `if not texto:`, `if cantidad:`… y evita `== True`.
- Nunca compares `float` con `==`: usa `math.isclose`.
- En Python no se puede asignar con `=` dentro de un `if` (adiós al bug de C). Si lo necesitas, `:=` con paréntesis.
- La expresión condicional `a if cond else b` es para **elegir valores**, no acciones. Con más de dos opciones, mejor `if`/`elif`.
- `match`: sin *fall-through*, `|` para varias opciones, `_` como comodín y guardas con `if`.
- En `match`, **un nombre suelto captura**, no compara. `True`/`False`/`None` se comparan con `is`.
- `x = condición` en lugar de `if condición: x = True else: x = False`.
