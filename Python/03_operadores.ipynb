## 1. Operadores aritméticos

| Operador | Operación | Ejemplo | Resultado |
| --- | --- | --- | --- |
| `+` | Suma | `7 + 2` | `9` |
| `-` | Resta | `7 - 2` | `5` |
| `*` | Multiplicación | `7 * 2` | `14` |
| `/` | División **real** | `7 / 2` | `3.5` |
| `//` | División **entera** (redondea hacia abajo) | `7 // 2` | `3` |
| `%` | Resto (módulo) | `7 % 2` | `1` |
| `**` | Potencia | `7 ** 2` | `49` |

Diferencia clave con C o Java: en Python 3, **`/` siempre devuelve `float`**, incluso entre enteros. Para la división entera existe `//`.

````python
print(7 / 2, type(7 / 2))
print(8 / 2, type(8 / 2))       # aunque sea exacta, es float
print(7 // 2, 7 % 2)
print(divmod(7, 2))             # cociente y resto a la vez
print(2 ** 10, 2 ** 0.5, 2 ** -1)
````

## 2. Asignación compuesta

`x += 1` es la forma abreviada de `x = x + 1`. Existe para todos los operadores aritméticos y bit a bit: `-=`, `*=`, `/=`, `//=`, `%=`, `**=`, `&=`, `|=`…

Con los tipos que conocemos (números y cadenas, que son inmutables), `x += 1` crea un objeto nuevo y mueve la etiqueta, igual que `x = x + 1`. Con las colecciones veremos que hay matices.

````python
contador = 10
contador += 5
contador *= 2
contador //= 3
print(contador)

texto = "Hola"
texto += " mundo"
print(texto)
````

## 4. Operadores lógicos: `and`, `or`, `not`

Dos características fundamentales:

1. **Evaluación en cortocircuito**: `and` deja de evaluar en cuanto encuentra un valor falso, `or` en cuanto encuentra uno verdadero.
2. **`and` y `or` no devuelven `True`/`False`**: devuelven **uno de sus operandos**.
   - `a or b` → devuelve `a` si es verdadero; si no, `b`.
   - `a and b` → devuelve `a` si es falso; si no, `b`.
   - `not` sí devuelve siempre un `bool`.

````python
print(True and False, True or False, not True)

print("hola" or "adiós")
print("" or "por defecto")
print(0 and 100)
print(5 and 100)
print(not "texto")
````

## 6. Operadores bit a bit

Trabajan sobre la representación binaria de los enteros. Se usan para permisos, flags, protocolos de red, máscaras…

| Operador | Operación | Ejemplo (`5` = `0b101`, `3` = `0b011`) |
| --- | --- | --- |
| `&` | AND | `5 & 3` → `1` (`0b001`) |
| `\|` | OR | `5 \| 3` → `7` (`0b111`) |
| `^` | XOR | `5 ^ 3` → `6` (`0b110`) |
| `~` | NOT | `~5` → `-6` |
| `<<` | Desplazamiento izquierda | `1 << 4` → `16` |
| `>>` | Desplazamiento derecha | `16 >> 2` → `4` |

````python
a, b = 5, 3
print(f"{a:04b} & {b:04b} = {a & b:04b}")
print(f"{a:04b} | {b:04b} = {a | b:04b}")
print(f"{a:04b} ^ {b:04b} = {a ^ b:04b}")
print(1 << 4, 16 >> 2)
````

## 7. Precedencia de operadores

De **mayor** a **menor** prioridad (los más habituales):

| Prioridad | Operadores |
| --- | --- |
| 1 | `()` paréntesis |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` (unarios) |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `in`, `not in`, `is`, `is not` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |
| 14 | `x if c else y` (expresión condicional) |
| 15 | `:=` (operador morsa, lo veremos en el tema 04) |

**Consejo práctico:** no hace falta memorizarla. Ante la duda, **pon paréntesis**: el código es más legible y evitas errores como los de los retos 1.4, 4.3 y 6.3.

## 8. Los operadores dependen del tipo

Un mismo operador hace cosas distintas según el tipo de los operandos. Cada tipo define su comportamiento mediante **métodos especiales** (`__add__`, `__mul__`, `__eq__`…); lo veremos en el tema de clases.

````python
print("ab" + "cd")
print("ab" * 3)
print("-" * 30)
print(f"{'TÍTULO':^30}")
print("-" * 30)
````

## ✅ Resumen

- `/` siempre devuelve `float`; `//` redondea **hacia abajo** (no hacia cero) y `%` tiene el signo del divisor.
- `**` tiene más prioridad que el `-` unario y se asocia por la derecha.
- No existe `++`: `++x` es `+(+x)` y no incrementa nada. Usa `x += 1`.
- Las comparaciones se **encadenan**: `a < b < c`.
- Las cadenas se comparan por código Unicode: `"10" < "9"` y `"Z" < "a"`.
- `and`/`or` hacen **cortocircuito** y **devuelven un operando**, no necesariamente un `bool`.
- `x == "a" or "b"` siempre es verdadero; hay que repetir la comparación: `x == "a" or x == "b"`.
- `"" in texto` siempre es `True`.
- `&` no es `and`: tiene otra precedencia. En pandas, **paréntesis obligatorios**.
- Ante la duda con la precedencia, **pon paréntesis**.
